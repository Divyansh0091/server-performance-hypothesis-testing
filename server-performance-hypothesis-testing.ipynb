```python
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
import scipy.stats as stats 
team_A = ([40,65,25,30,14,25,65,20,38,46,20,85,95,45,20,89,59,54,24,56])
team_B = ([65,58,45,25,45,65,25,98,65,25,35,85,45,75,65,25,35,25,25,45])
```

```python
mean_1 = np.mean(team_A)
mean_2 = np.mean(team_B) 
std_1 = np.std(team_A) 
std_2 = np.std(team_B) 
n = 20
```

```python
critical_val = stats.t.ppf(1-0.05,19) 
p_val = (mean_1/mean_2)/np.sqrt((std_1/n)+(std_2/n))
print(f"Critical-Val {critical_val}") 
print(f"P-Val {p_val}") 
if p_val < critical_val: 
    print(f"Result: {"Hypothesis Rejected"}") 
else:
    print(f"Result: {"Hypthesis not Rejected"}") 
```

```python
plt.figure(figsize=(10,6))
sns.kdeplot(data=team_A, fill=True, color="#e74c3c", label='Team-A', linewidth=2.5, alpha=0.3)
sns.kdeplot(data=team_B, fill=True, color="#2ecc71", label='Team-B', linewidth=2.5, alpha=0.3)
plt.axvline(np.mean(team_A),color='#c0392b', linestyle='--', linewidth=2)
plt.axvline(np.mean(team_B),color='#27ae60', linestyle='--', linewidth=2)
plt.title('Server Performance Distribution: Old vs New Server', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Server Response Time / Speed Metrics', fontsize=12, labelpad=10)
plt.ylabel('Density', fontsize=12, labelpad=10)
plt.legend(loc='upper right', fontsize=11, frameon=True, facecolor='white', edgecolor='none')
plt.tight_layout()
plt.show()
```
